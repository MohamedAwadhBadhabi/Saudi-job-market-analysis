# Statistical Analysis

## Introduction

This notebook applies statistical methods to investigate relationships between key variables in the Saudi job market dataset.

The goal is not only to describe the data but also to determine whether observed patterns are statistically significant.

The analysis includes:
- Chi-Square Test of Independence
- Statistical interpretation
- Business implications

```python
import pandas as pd
import numpy as np

from scipy.stats import chi2_contingency, f_oneway
```

## Statistical Question 1

### Is there a statistically significant association between industry and gender requirements in job postings?

### Hypotheses

**Null Hypothesis (H₀):**
There is no association between industry and gender requirements.

**Alternative Hypothesis (H₁):**
There is a significant association between industry and gender requirements.

```python
df = pd.read_csv("data/processed/Jadarat_cleaned.csv")

df.head()
```

```python
contingency_table = pd.crosstab(
    df["eco_activity"],
    df["gender"]
)

contingency_table.head()
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")
```

```python


n = contingency_table.to_numpy().sum()

cramers_v = np.sqrt(chi2 / (n * (min(contingency_table.shape) - 1)))

print(f"Cramér's V: {cramers_v:.4f}")
```

### Interpretation

A Chi-Square Test of Independence was conducted to examine the relationship between industry and gender requirements in job postings.

The results showed a statistically significant association between the two variables (χ² = 1210.24, p < 0.001). Therefore, the null hypothesis was rejected.

To assess the strength of this relationship, Cramér's V was calculated. The result (V = 0.6622) indicates a **strong association** between industry and gender requirements.

This suggests that gender preferences in job advertisements vary substantially across different industries.

### Business Insight

The analysis indicates that hiring preferences are strongly associated with industry type.

Some industries predominantly advertise positions for male applicants, while others recruit female applicants or accept both genders. These differences are statistically significant and represent meaningful hiring patterns rather than random variation.

Understanding these patterns can help policymakers monitor labor market trends and assist organizations in evaluating recruitment practices.

## Statistical Question 2

### Is there a statistically significant association between region and industry in Saudi Arabia?

```python
top20_industries = df["eco_activity"].value_counts().head(20).index 

df["industry_grouped"] = df["eco_activity"].apply(
    lambda x: x if x in top20_industries else "Other"
)
```

```python
df["industry_grouped"].value_counts()
```

```python
contingency_table = pd.crosstab(
    df["region"],
    df["industry_grouped"]
)

contingency_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")
```

```python
n = contingency_table.to_numpy().sum()

cramers_v = np.sqrt(
    chi2 / (n * (min(contingency_table.shape) - 1))
)

print(f"Cramér's V: {cramers_v:.4f}")
```

### Interpretation

A Chi-Square Test of Independence was conducted to examine the relationship between region and industry.

The results revealed a statistically significant association between the two variables (χ² = 1058.18, p < 0.001). Therefore, the null hypothesis was rejected.

Cramér's V was calculated to measure the strength of the association. The result (V = 0.2353) indicates a **weak association** between region and industry.

This suggests that although the distribution of industries varies across regions, the relationship is relatively weak.

### Business Insight

The analysis suggests that industry distribution differs across Saudi regions, but the differences are relatively modest.

This indicates that many industries are present in multiple regions rather than being concentrated in only a few locations. Organizations and policymakers can use this information to understand regional economic diversity and identify areas for future investment or workforce development.

## Statistical Question 3

### Business Question

Do different industries require different average years of experience?

```python
df["exper"].value_counts().sort_index()
```

```python
df["exper_years"] = df["exper"].apply(
    lambda x: int(x.split()[0])
) 
```

```python
df.to_csv(
    "data/processed/Jadarat_cleaned.csv",
    index=False
)
```

```python
df[["exper", "exper_years"]].head(10)
```

```python
df.groupby("industry_grouped")["exper_years"].mean().sort_values(ascending=False)
```

```python
groups = [
    group["exper_years"].values
    for _, group in df.groupby("industry_grouped")
]

len(groups)
```

```python
f_statistic, p_value = f_oneway(*groups)

print(f"F-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

### Interpretation

A one-way ANOVA was conducted to determine whether the average years of experience required differ across industries.

The results showed a statistically significant difference in average years of experience among industries (F = 5.79, p < 0.001).

Therefore, the null hypothesis was rejected, indicating that at least one industry requires a significantly different average level of experience compared to the others.

### Business Insight

The findings suggest that experience requirements vary across industries in Saudi Arabia.

Some industries tend to seek more experienced candidates, while others are more open to entry-level applicants. This information can help job seekers identify industries that match their experience level and assist employers and policymakers in understanding workforce demand across sectors.
