```python
import matplotlib.pyplot as plt

top_jobs = df["job_title"].value_counts().head(10)

plt.figure(figsize=(10,6))

bars = plt.bar(top_jobs.index, top_jobs.values)

plt.title("Top 10 Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")

plt.xticks(rotation=90)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        int(height),
        ha="center",
        va="bottom"
    )
plt.ylim(0, 110)
plt.show()
```

```python
plt.ylim(0, 110)
```

### Insight

Salesperson (بائع) is the most frequently advertised job title with 98 postings, followed by Accountant (محاسب) with 89 postings.

Sales, accounting, and marketing roles dominate the top job postings in this dataset.

```python
import matplotlib.pyplot as plt

top_cities = df["city"].value_counts().head(10)

plt.figure(figsize=(10,6))

bars = plt.bar(top_cities.index, top_cities.values)

plt.title("Top 10 Cities")
plt.xlabel("City")
plt.ylabel("Number of Jobs")

plt.xticks(rotation=90)
plt.ylim(0, 600)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        int(height),
        ha="center",
        va="bottom"
    )

plt.show()
```

## Insight

Riyadh has the highest number of job postings (523), followed by Jeddah (245). This indicates that job opportunities are concentrated in major urban areas, particularly the capital city.

```python
top_regions = df["region"].value_counts().head(10)

plt.figure(figsize=(10,6))

bars = plt.bar(top_regions.index, top_regions.values)

plt.title("Top 10 Regions by Job Postings")
plt.xlabel("Region")
plt.ylabel("Number of Job Postings")

plt.xticks(rotation=90)
plt.ylim(0, 700)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 3,
        int(height),
        ha="center",
        va="bottom"
    )

plt.show()
```

## Insight

Riyadh has the highest number of job postings (622), followed by Makkah (369) and the Eastern Province (219). This indicates that employment opportunities are concentrated in the central and western regions of Saudi Arabia.

```python
# 1. تجهيز البيانات
gender_counts = df["gender"].value_counts()

# 2. إنشاء الشكل
plt.figure(figsize=(6,6))

# 3. رسم الرسم
plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%"
)

# 4. العنوان
plt.title("Gender Distribution")

# 5. العرض
plt.axis("equal")
plt.show()
```

```python
print(gender_counts)
```

## Insight

The gender distribution in the dataset is relatively balanced. Jobs open to both genders represent the largest share (39.9%), followed by male-only positions (32.7%) and female-only positions (27.4%). This indicates that many job opportunities are available to both men and women, although some positions remain gender-specific.

```python
# 1. تجهيز البيانات
contract_counts = df["contract"].value_counts()

# 2. إنشاء الشكل
plt.figure(figsize=(6,6))

# 3. رسم الرسم
plt.pie(
    contract_counts.values,
    labels=contract_counts.index,
    autopct="%1.1f%%"
)

# 4. العنوان
plt.title("Contract Type Distribution")

# 5. العرض
plt.axis("equal")
plt.show()
```

## Insight

Full-time positions dominate the Saudi job market, while remote work opportunities represent only a small portion of the job postings. This suggests that employers primarily prefer traditional full-time employment over remote work.
